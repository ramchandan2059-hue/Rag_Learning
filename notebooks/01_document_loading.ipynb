### TEXT FILE LOADER

```python
from langchain_community.document_loaders import TextLoader
```

```python
loader = TextLoader(file_path="../data/abc.txt")
data = loader.load()
```

```python
len(data)
```

```python
print(data[0].page_content)
```

```python
print(data[0].metadata)
```

### Web Based Loader

```python
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader(web_path="https://vector.techsimplus.com/")
```

```python
data = loader.load()
```

```python
len(data)
```

```python
data[0].metadata
```

```python
print(data[0].page_content)
```
